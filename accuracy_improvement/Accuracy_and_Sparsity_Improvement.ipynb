# Accuracy and Sparsity Improvement

Inspect the completed study without retraining. Experiment 1 addresses approximately preserved binary accuracy; Experiment 2 addresses joint macro F1 and sparsity improvement. Both use the same validation-selected models. See the root README for fresh-training commands and limitations.

```python
from pathlib import Path
import csv
from IPython.display import display, Image
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "accuracy_improvement/results/summary.csv").is_file())
RESULTS = ROOT / "accuracy_improvement/results"
with (RESULTS / "summary.csv").open() as f:
    rows = list(csv.DictReader(f))
for row in rows:
    if row["method"] in ("fixed_set", "refined_policy"):
        print(row["method"], "accuracy", row["accuracy_mean"], "macro F1", row["macro_f1_mean"], "bytes", row["compact_weight_bytes_mean"])
```

```python
display(Image(filename=str(RESULTS / "training-curves.png")))
display(Image(filename=str(RESULTS / "warm-start-curves.png")))
display(Image(filename=str(RESULTS / "accuracy-size-tradeoff.png")))
```

The saved baseline and compressed policy round to the same binary accuracy of 91.53%, but their exact difference is −0.0045 percentage points. Macro F1 improves by 0.1831 percentage points and compact weight bytes decrease by 17.36%. Three seeds do not establish statistical superiority or equivalence. Refinement uses 25 additional epochs after a 50-epoch source.
